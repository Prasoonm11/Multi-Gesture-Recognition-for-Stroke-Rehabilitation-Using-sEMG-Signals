```python
import matplotlib.pyplot as plt
import numpy as np

def plot_trial_channels_fixed(grasp_label, trial_index=0):
    """
    Fixed plotting function that handles the actual data structure
    """
    # Check if key exists
    if grasp_label not in combined_grasp_dict:
        print(f"Error: '{grasp_label}' not found in combined_grasp_dict")
        print(f"Available keys: {list(combined_grasp_dict.keys())}")
        return

    trials = combined_grasp_dict[grasp_label]

    # Check if trials list is not empty
    if len(trials) == 0:
        print(f"Error: No trials found for '{grasp_label}'")
        return

    # Check if trial_index is valid
    if trial_index >= len(trials):
        print(f"Error: trial_index {trial_index} out of range. Available: 0-{len(trials)-1}")
        return

    trial_data = trials[trial_index]

    print(f"Plotting {grasp_label} - Trial {trial_index}")
    print(f"Original data shape: {trial_data.shape}")

    # IMPORTANT: Transpose the data from (channels, time_samples) to (time_samples, channels)
    if trial_data.shape[0] == 16:  # If first dimension is 16 (channels)
        trial_data = trial_data.T  # Transpose to (time_samples, channels)
        print(f"Transposed data shape: {trial_data.shape}")

    num_channels = trial_data.shape[1]
    num_time_samples = trial_data.shape[0]

    print(f"Time samples: {num_time_samples}, Channels: {num_channels}")

    # Create 4x4 subplot layout for 16 channels
    plt.figure(figsize=(15, 10))

    for channel in range(min(16, num_channels)):
        plt.subplot(4, 4, channel + 1)
        plt.plot(trial_data[:, channel])
        plt.title(f'Channel {channel + 1}')
        plt.xlabel('Time samples')
        plt.ylabel('sEMG Signal')

    plt.suptitle(f'{grasp_label} - Trial {trial_index + 1}')
    plt.tight_layout()
    plt.show()

# Function to plot all grasp types
def plot_all_grasp_types():
    """
    Plot first trial of each available grasp type
    """
    valid_keys = ['Pointer', 'Power', 'Open', 'Tripod', 'Lateral', 'Rest']

    for grasp_label in valid_keys:
        print(f"\n{'='*50}")
        print(f"Plotting {grasp_label}")
        print('='*50)
        plot_trial_channels_fixed(grasp_label, 0)
```

```python
import numpy as np
from scipy.fft import fft, fftfreq
import matplotlib.pyplot as plt

# Simple FFT conversion function
def emg_to_fft(combined_grasp_dict, fs=2000, n_freq_bins=500):
    """Convert EMG data to FFT with sampling frequency fs and fixed frequency bins"""

    fft_dict = {}

    for grasp_label, trials in combined_grasp_dict.items():
        print(f"Converting {grasp_label}...")

        fft_trials = []
        for trial in trials:
            # Ensure shape is (16, time_samples)
            if trial.shape[0] != 16:
                trial = trial.T

            # Apply FFT to each channel
            fft_trial = np.abs(fft(trial, axis=1))  # FFT along time axis

            # Keep only positive frequencies (first half)
            fft_trial = fft_trial[:, :fft_trial.shape[1]//2]

            # Remove DC component (0 Hz) by setting first bin to 0
            fft_trial[:, 0] = 0

            # Downsample to fixed number of frequency bins
            if fft_trial.shape[1] > n_freq_bins:
                # Downsample by taking every nth point
                step = fft_trial.shape[1] // n_freq_bins
                fft_trial = fft_trial[:, ::step][:, :n_freq_bins]
            elif fft_trial.shape[1] < n_freq_bins:
                # Pad with zeros if needed
                padding = n_freq_bins - fft_trial.shape[1]
                fft_trial = np.pad(fft_trial, ((0, 0), (0, padding)), mode='constant')

            fft_trials.append(fft_trial)

        fft_dict[grasp_label] = fft_trials

    # Create frequency axis for fixed bins
    frequencies = np.linspace(0, fs/2, n_freq_bins)

    return fft_dict, frequencies

# Convert your data
fft_data, freq_axis = emg_to_fft(combined_grasp_dict, fs=2000, n_freq_bins=500)

# Check shapes
print("\\nFFT Data Shapes:")
for grasp, trials in fft_data.items():
    print(f"{grasp}: {len(trials)} trials, shape: {trials[0].shape}")

print(f"\\nFrequency range: {freq_axis[0]:.1f} to {freq_axis[-1]:.1f} Hz")

# Plot FFT spectra
def plot_fft_comparison(fft_data, frequencies):
    """Plot FFT comparison for all grasp types"""

    fig, axes = plt.subplots(2, 3, figsize=(15, 10))
    axes = axes.flatten()

    colors = ['blue', 'red', 'green', 'orange', 'purple', 'brown']

    for i, (grasp_label, trials) in enumerate(fft_data.items()):
        ax = axes[i]

        # Average across all trials and channels
        avg_spectrum = np.mean([np.mean(trial, axis=0) for trial in trials], axis=0)

        # Plot average spectrum
        ax.plot(frequencies, avg_spectrum, color=colors[i], linewidth=2)
        ax.set_title(f'{grasp_label}', fontsize=14, fontweight='bold')
        ax.set_xlabel('Frequency (Hz)')
        ax.set_ylabel('Magnitude')
        ax.grid(True, alpha=0.3)
        ax.set_xlim(10, 500)  # Start from 10 Hz to avoid DC component

        # Add some styling
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)

    plt.tight_layout()
    plt.suptitle('FFT Spectra Comparison Across Grasp Types', fontsize=16, y=1.02)
    plt.show()

# Plot single channel comparison
def plot_single_channel_fft(fft_data, frequencies, channel=0):
    """Plot FFT for a single channel across all grasp types"""

    plt.figure(figsize=(12, 8))

    colors = ['blue', 'red', 'green', 'orange', 'purple', 'brown']

    for i, (grasp_label, trials) in enumerate(fft_data.items()):
        # Average spectrum for this channel across all trials
        channel_spectra = [trial[channel, :] for trial in trials]
        avg_spectrum = np.mean(channel_spectra, axis=0)

        plt.plot(frequencies, avg_spectrum, label=grasp_label,
                color=colors[i], linewidth=2, alpha=0.8)

    plt.xlabel('Frequency (Hz)', fontsize=12)
    plt.ylabel('Magnitude', fontsize=12)
    plt.title(f'FFT Comparison - Channel {channel+1}', fontsize=14, fontweight='bold')
    plt.legend(fontsize=10)
    plt.grid(True, alpha=0.3)
    plt.xlim(10, 500)

    # Style the plot
    plt.gca().spines['top'].set_visible(False)
    plt.gca().spines['right'].set_visible(False)

    plt.tight_layout()
    plt.show()

# Create the plots
print("\\nGenerating FFT plots...")

# Plot 1: Individual subplots for each grasp type
plot_fft_comparison(fft_data, freq_axis)

# Plot 2: Overlay comparison for channel 0
plot_single_channel_fft(fft_data, freq_axis, channel=0)

# Quick stats
print("\\nQuick FFT Statistics:")
for grasp_label, trials in fft_data.items():
    # Average power in EMG band (20-250 Hz)
    # Find frequency indices for 20-250 Hz range
    freq_start_idx = int(20 * len(freq_axis) / (freq_axis[-1]))  # 20 Hz index
    freq_end_idx = int(250 * len(freq_axis) / (freq_axis[-1]))   # 250 Hz index

    # Calculate average power in EMG band
    emg_powers = []
    for trial in trials:
        # Average across channels and frequency bins in EMG range
        emg_power = np.mean(trial[:, freq_start_idx:freq_end_idx])
        emg_powers.append(emg_power)

    avg_emg_power = np.mean(emg_powers)
    print(f"{grasp_label}: Average EMG power (20-250 Hz) = {avg_emg_power:.4f}")

    # Also show peak frequency for each grasp type
    avg_spectrum = np.mean([np.mean(trial, axis=0) for trial in trials], axis=0)
    peak_freq_idx = np.argmax(avg_spectrum[freq_start_idx:freq_end_idx]) + freq_start_idx
    peak_freq = freq_axis[peak_freq_idx]
    print(f"{grasp_label}: Peak frequency = {peak_freq:.1f} Hz")
```

```python
import numpy as np
from sklearn.model_selection import train_test_split

def split_fft_data(fft_data):

    train_data = {}
    test_data = {}

    # Get valid grasp types (skip empty ones and invalid keys)
    valid_grasps = []
    for grasp_name in fft_data.keys():
        if isinstance(grasp_name, str) and len(fft_data[grasp_name]) > 0:
            valid_grasps.append(grasp_name)

    print(f"Found {len(valid_grasps)} grasp types: {valid_grasps}")
    print()

    # Split each grasp type
    for grasp_name in valid_grasps:
        # Get all trials for this grasp
        all_trials = fft_data[grasp_name]
        total_trials = len(all_trials)

        print(f"Splitting {grasp_name}:")
        print(f"  Total trials: {total_trials}")

        # Validate FFT data shape
        if total_trials > 0:
            sample_shape = all_trials[0].shape
            print(f"  FFT data shape: {sample_shape} (channels x frequency_bins)")

        # Create list of trial numbers [0, 1, 2, ..., n-1]
        trial_numbers = list(range(total_trials))

        # Split trial numbers into train (70%) and test (30%)
        train_numbers, test_numbers = train_test_split(
            trial_numbers,
            test_size=0.3,  # 30% for testing
            random_state=42,  # for reproducible results
            stratify=None  # No stratification needed for regression/continuous data
        )

        # Get the actual trials using the split numbers
        train_trials = [all_trials[i] for i in train_numbers]
        test_trials = [all_trials[i] for i in test_numbers]

        # Store in our result dictionaries
        train_data[grasp_name] = train_trials
        test_data[grasp_name] = test_trials

        print(f"  Training: {len(train_trials)} trials (70%)")
        print(f"  Testing: {len(test_trials)} trials (30%)")
        print()

    return train_data, test_data

def show_split_results(train_data, test_data):

    print("FFT DATA SPLIT SUMMARY")

    total_train = 0
    total_test = 0

    for grasp_name in train_data.keys():
        train_count = len(train_data[grasp_name])
        test_count = len(test_data[grasp_name])

        total_train += train_count
        total_test += test_count

        print(f"{grasp_name}:")
        print(f"  Training: {train_count} trials")
        print(f"  Testing:  {test_count} trials")

        # Show data shape info
        if train_count > 0:
            shape = train_data[grasp_name][0].shape
            print(f"  Data shape: {shape} (channels x frequency_bins)")
        print()

    print(f"TOTAL:")
    print(f"  Training trials: {total_train}")
    print(f"  Testing trials:  {total_test}")
    print(f"  Total trials:    {total_train + total_test}")
    print(f"  Split ratio:     {total_train/(total_train + total_test)*100:.1f}% train, {total_test/(total_train + total_test)*100:.1f}% test")

def validate_fft_split(train_data, test_data, freq_axis):
    """
    Validate the FFT data split and show additional statistics
    """
    print("\nFFT DATA VALIDATION:")

    for grasp_name in train_data.keys():
        train_trials = train_data[grasp_name]
        test_trials = test_data[grasp_name]

        print(f"\n{grasp_name}:")

        # Check shapes consistency
        if len(train_trials) > 0 and len(test_trials) > 0:
            train_shape = train_trials[0].shape
            test_shape = test_trials[0].shape

            if train_shape == test_shape:
                print(f"  Shape consistency: {train_shape}")
            else:
                print(f"  Shape mismatch: train {train_shape}, test {test_shape}")

        # Check for data integrity (no NaN or inf values)
        train_valid = all(np.isfinite(trial).all() for trial in train_trials)
        test_valid = all(np.isfinite(trial).all() for trial in test_trials)

        print(f"  Training data valid: {train_valid}")
        print(f"  Testing data valid: {test_valid}")

        # Show some statistics
        if len(train_trials) > 0:
            train_mean = np.mean([np.mean(trial) for trial in train_trials])
            test_mean = np.mean([np.mean(trial) for trial in test_trials])
            print(f"  Average magnitude - Train: {train_mean:.4f}, Test: {test_mean:.4f}")


# Step 1: Split your FFT data
print("Splitting FFT data into train and test sets...")
train_fft_data, test_fft_data = split_fft_data(fft_data)

# Step 2: Show the results
show_split_results(train_fft_data, test_fft_data)

# Step 3: Validate the split (optional but recommended)
validate_fft_split(train_fft_data, test_fft_data, freq_axis)

# Step 4: Your data is now ready!
print("YOUR FFT DATA IS NOW SPLIT AND READY TO USE!")

# Optional: Save the split data
# np.save('train_fft_data.npy', train_fft_data)
# np.save('test_fft_data.npy', test_fft_data)
# print("\n✓ Split data saved to files!")
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
import snntorch as snn
from snntorch import spikegen

class FFTRateEncoder:

    def __init__(self, num_steps=30, gain=2, dtype=torch.float32):
        """
        Initialize the FFT Rate Encoder
            num_steps (int): Number of time steps for spike encoding
            gain (float): Gain factor for rate encoding (controls spike probability)
            dtype: PyTorch data type for tensors
        """
        self.num_steps = num_steps
        self.gain = gain
        self.dtype = dtype
        self.scaler = MinMaxScaler()

    def preprocess_fft_data(self, train_fft_data, test_fft_data):
        """
        1. Normalize FFT magnitude data to [0, 1] range (required for rate encoding)
        2. Convert to PyTorch tensors
        3. Organize data for batch processing
        """

        print("=== FFT DATA PREPROCESSING ===")

        # Step 1: Collect all FFT data to fit the scaler
        print("1. Collecting all FFT data for normalization...")
        all_fft_data = []

        for grasp_type in train_fft_data.keys():
            for trial in train_fft_data[grasp_type]:
                # trial shape: (16 channels, 500 frequency_bins)
                # Flatten to (16 * 500,) for scaling
                all_fft_data.append(trial.flatten())

        # Combine all data and fit scaler
        combined_data = np.concatenate(all_fft_data)
        combined_data = combined_data.reshape(-1, 1)  # Reshape for sklearn

        print(f"   - Total FFT samples for scaling: {len(combined_data)}")
        print(f"   - FFT magnitude range: [{combined_data.min():.6f}, {combined_data.max():.6f}]")

        # Fit scaler to normalize data to [0, 1]
        self.scaler.fit(combined_data)

        # Step 2: Process training data
        print("\n 2. Processing training data...")
        train_processed = {}

        for grasp_type in train_fft_data.keys():
            processed_trials = []

            for trial in train_fft_data[grasp_type]:
                # Normalize the trial
                trial_flat = trial.flatten().reshape(-1, 1)
                trial_normalized = self.scaler.transform(trial_flat)
                trial_normalized = trial_normalized.reshape(trial.shape)

                # Convert to PyTorch tensor
                trial_tensor = torch.tensor(trial_normalized, dtype=self.dtype)
                processed_trials.append(trial_tensor)

            train_processed[grasp_type] = processed_trials
            print(f"   - {grasp_type}: {len(processed_trials)} trials processed")

        # Step 3: Process testing data
        print("\n 3. Processing testing data...")
        test_processed = {}

        for grasp_type in test_fft_data.keys():
            processed_trials = []

            for trial in test_fft_data[grasp_type]:
                # Normalize the trial
                trial_flat = trial.flatten().reshape(-1, 1)
                trial_normalized = self.scaler.transform(trial_flat)
                trial_normalized = trial_normalized.reshape(trial.shape)

                # Convert to PyTorch tensor
                trial_tensor = torch.tensor(trial_normalized, dtype=self.dtype)
                processed_trials.append(trial_tensor)

            test_processed[grasp_type] = processed_trials
            print(f"   - {grasp_type}: {len(processed_trials)} trials processed")

        print("\\n✓ FFT preprocessing completed!")
        return train_processed, test_processed

    def encode_single_trial(self, trial_data, verbose=False):
        """
        Apply rate encoding to a single FFT trial

        Args:
            trial_data (torch.Tensor): Single trial FFT data (16 channels, 500 freq_bins)
            verbose (bool): Print detailed information

        Returns:
            torch.Tensor: Spike train (num_steps, 16 channels, 500 freq_bins)
        """

        if verbose:
            print(f"   - Input shape: {trial_data.shape}")
            print(f"   - Input range: [{trial_data.min():.3f}, {trial_data.max():.3f}]")

        # Apply rate encoding using snnTorch
        # This creates spike trains based on FFT magnitude values
        spike_train = spikegen.rate(trial_data,
                                   num_steps=self.num_steps,
                                   gain=self.gain)

        if verbose:
            print(f"   - Output shape: {spike_train.shape}")
            print(f"   - Total spikes: {spike_train.sum().item()}")
            print(f"   - Spike rate: {spike_train.mean().item():.3f}")

        return spike_train

    def encode_all_data(self, train_processed, test_processed):
        """
        Apply rate encoding to all training and testing data

        Args:
            train_processed (dict): Preprocessed training data
            test_processed (dict): Preprocessed testing data

        Returns:
            tuple: (train_encoded, test_encoded) dictionaries
        """

        print("\\n=== RATE ENCODING ===")

        # Encode training data
        print("1. Encoding training data...")
        train_encoded = {}

        for grasp_type in train_processed.keys():
            encoded_trials = []

            for i, trial in enumerate(train_processed[grasp_type]):
                spike_train = self.encode_single_trial(trial, verbose=(i==0))
                encoded_trials.append(spike_train)

            train_encoded[grasp_type] = encoded_trials
            print(f"   - {grasp_type}: {len(encoded_trials)} trials encoded")

        # Encode testing data
        print("\n 2. Encoding testing data...")
        test_encoded = {}

        for grasp_type in test_processed.keys():
            encoded_trials = []

            for i, trial in enumerate(test_processed[grasp_type]):
                spike_train = self.encode_single_trial(trial, verbose=(i==0))
                encoded_trials.append(spike_train)

            test_encoded[grasp_type] = encoded_trials
            print(f"   - {grasp_type}: {len(encoded_trials)} trials encoded")

        print("\\n✓ Rate encoding completed!")
        return train_encoded, test_encoded

    def visualize_fft_encoding(self, original_data, encoded_data, grasp_type, trial_idx=0, channel_idx=0):
        """
        Visualize the FFT rate encoding results

        Args:
            original_data (dict): Original preprocessed FFT data
            encoded_data (dict): Rate-encoded spike data
            grasp_type (str): Grasp type to visualize
            trial_idx (int): Trial index to visualize
            channel_idx (int): Channel index to visualize
        """

        print(f"\\n=== VISUALIZATION: {grasp_type} (Trial {trial_idx}, Channel {channel_idx}) ===")

        # Get original and encoded data
        original_trial = original_data[grasp_type][trial_idx]
        encoded_trial = encoded_data[grasp_type][trial_idx]

        # Extract single channel data
        original_channel = original_trial[channel_idx, :]  # Shape: (500,)
        encoded_channel = encoded_trial[:, channel_idx, :]  # Shape: (num_steps, 500)

        # Create frequency axis for plotting
        freq_axis = np.linspace(0, 1000, 500)  # 0 to 1000 Hz, 500 bins

        # Create visualization
        fig, axes = plt.subplots(3, 1, figsize=(15, 12))

        # Plot 1: Original FFT magnitude spectrum
        axes[0].plot(freq_axis, original_channel.numpy(), 'b-', linewidth=1)
        axes[0].set_title(f'Original FFT Magnitude - {grasp_type} (Channel {channel_idx})')
        axes[0].set_xlabel('Frequency (Hz)')
        axes[0].set_ylabel('Normalized Magnitude')
        axes[0].grid(True, alpha=0.3)
        axes[0].set_xlim(10, 500)  # Focus on EMG-relevant frequencies

        # Plot 2: Spike raster plot across frequency bins
        spike_times, spike_freqs = torch.where(encoded_channel == 1)
        if len(spike_times) > 0:
            axes[1].scatter(freq_axis[spike_freqs.numpy()], spike_times.numpy(),
                           s=1, c='red', alpha=0.6)
        axes[1].set_title(f'Spike Raster Plot - {grasp_type} (Channel {channel_idx})')
        axes[1].set_xlabel('Frequency (Hz)')
        axes[1].set_ylabel('Time Step')
        axes[1].set_xlim(10, 500)
        axes[1].set_ylim(0, self.num_steps)
        axes[1].grid(True, alpha=0.3)

        # Plot 3: Average spike rate across frequency bins
        avg_spike_rate = encoded_channel.mean(dim=0)  # Average across time steps
        axes[2].plot(freq_axis, avg_spike_rate.numpy(), 'g-', linewidth=2)
        axes[2].set_title(f'Average Spike Rate per Frequency - {grasp_type} (Channel {channel_idx})')
        axes[2].set_xlabel('Frequency (Hz)')
        axes[2].set_ylabel('Average Spike Rate')
        axes[2].set_xlim(10, 500)
        axes[2].grid(True, alpha=0.3)

        plt.tight_layout()
        plt.show()

        # Print statistics
        total_spikes = encoded_channel.sum().item()
        spike_rate = encoded_channel.mean().item()

        print(f"Statistics for {grasp_type} (Trial {trial_idx}, Channel {channel_idx}):")
        print(f"  - Total spikes: {total_spikes}")
        print(f"  - Spike rate: {spike_rate:.3f}")
        print(f"  - Original FFT magnitude range: [{original_channel.min():.3f}, {original_channel.max():.3f}]")
        print(f"  - Peak frequency: {freq_axis[original_channel.argmax()]:.1f} Hz")

    def compare_grasp_types(self, encoded_data, num_trials=5):
        """
        Compare spike patterns across different grasp types

        Args:
            encoded_data (dict): Rate-encoded spike data
            num_trials (int): Number of trials to analyze per grasp type
        """

        print("\\n=== GRASP TYPE COMPARISON ===")

        # Calculate statistics for each grasp type
        grasp_stats = {}

        for grasp_type in encoded_data.keys():
            spike_rates = []
            total_spikes = []

            for trial_idx in range(min(num_trials, len(encoded_data[grasp_type]))):
                trial_data = encoded_data[grasp_type][trial_idx]
                spike_rate = trial_data.mean().item()
                total_spike = trial_data.sum().item()

                spike_rates.append(spike_rate)
                total_spikes.append(total_spike)

            grasp_stats[grasp_type] = {
                'avg_spike_rate': np.mean(spike_rates),
                'std_spike_rate': np.std(spike_rates),
                'avg_total_spikes': np.mean(total_spikes),
                'std_total_spikes': np.std(total_spikes)
            }

        # Create comparison plot
        fig, axes = plt.subplots(1, 2, figsize=(15, 6))

        # Plot 1: Average spike rates
        grasp_names = list(grasp_stats.keys())
        avg_rates = [grasp_stats[name]['avg_spike_rate'] for name in grasp_names]
        std_rates = [grasp_stats[name]['std_spike_rate'] for name in grasp_names]

        axes[0].bar(grasp_names, avg_rates, yerr=std_rates, capsize=5, alpha=0.7)
        axes[0].set_title('Average Spike Rate by Grasp Type')
        axes[0].set_ylabel('Spike Rate')
        axes[0].tick_params(axis='x', rotation=45)
        axes[0].grid(True, alpha=0.3)

        # Plot 2: Total spikes
        avg_totals = [grasp_stats[name]['avg_total_spikes'] for name in grasp_names]
        std_totals = [grasp_stats[name]['std_total_spikes'] for name in grasp_names]

        axes[1].bar(grasp_names, avg_totals, yerr=std_totals, capsize=5, alpha=0.7)
        axes[1].set_title('Average Total Spikes by Grasp Type')
        axes[1].set_ylabel('Total Spikes')
        axes[1].tick_params(axis='x', rotation=45)
        axes[1].grid(True, alpha=0.3)

        plt.tight_layout()
        plt.show()

        # Print detailed statistics
        print("\n Detailed Statistics:")
        for grasp_type in grasp_names:
            stats = grasp_stats[grasp_type]
            print(f"\\n{grasp_type}:")
            print(f"  - Avg spike rate: {stats['avg_spike_rate']:.3f} ± {stats['std_spike_rate']:.3f}")
            print(f"  - Avg total spikes: {stats['avg_total_spikes']:.0f} ± {stats['std_total_spikes']:.0f}")

# MAIN EXECUTION CODE

def main():

    print("FFT Rate Encoding for Spiking Neural Networks")

    # Initialize the encoder
    encoder = FFTRateEncoder(num_steps=30, gain=2)

    # Step 1: Preprocess the FFT data
    train_processed, test_processed = encoder.preprocess_fft_data(train_fft_data, test_fft_data)

    # Step 2: Apply rate encoding
    train_encoded, test_encoded = encoder.encode_all_data(train_processed, test_processed)

    # Step 3: Visualize the encoding for one example
    # Get the first available grasp type for visualization
    first_grasp = list(train_encoded.keys())[0]
    encoder.visualize_fft_encoding(train_processed, train_encoded,
                                 grasp_type=first_grasp, trial_idx=0, channel_idx=0)

    # Step 4: Compare different grasp types
    encoder.compare_grasp_types(train_encoded, num_trials=5)

    print("\n  FFT rate encoding completed successfully!")

    return train_encoded, test_encoded

# Run the main function
if __name__ == "__main__":
    train_encoded, test_encoded = main()

def analyze_fft_encoding_quality(encoded_data, grasp_type, trial_idx=0):
    """
    Analyze the quality of FFT rate encoding for a specific trial

    Args:
        encoded_data (dict): Rate-encoded spike data
        grasp_type (str): Grasp type to analyze
        trial_idx (int): Trial index to analyze
    """

    trial = encoded_data[grasp_type][trial_idx]

    print(f"\\n=== FFT ENCODING QUALITY ANALYSIS ===")
    print(f"Grasp Type: {grasp_type}, Trial: {trial_idx}")
    print(f"Encoded shape: {trial.shape}")

    # Calculate statistics
    total_spikes = trial.sum().item()
    total_possible = trial.numel()
    spike_rate = trial.mean().item()

    print(f"Total spikes: {total_spikes}")
    print(f"Total possible spikes: {total_possible}")
    print(f"Overall spike rate: {spike_rate:.3f}")

    # Channel-wise analysis
    channel_spike_rates = trial.mean(dim=(0, 2))  # Average over time and frequency
    print(f"Channel spike rates (mean): {channel_spike_rates.mean().item():.3f}")
    print(f"Channel spike rates (std): {channel_spike_rates.std().item():.3f}")

    # Temporal analysis
    temporal_spike_rates = trial.mean(dim=(1, 2))  # Average over channels and frequency
    print(f"Temporal spike rates (first 10 steps): {temporal_spike_rates[:10]}")

    # Frequency analysis
    frequency_spike_rates = trial.mean(dim=(0, 1))  # Average over time and channels
    peak_freq_idx = frequency_spike_rates.argmax().item()
    print(f"Peak frequency bin: {peak_freq_idx} (corresponds to ~{peak_freq_idx*2} Hz)")

def save_encoded_fft_data(train_encoded, test_encoded, filename="fft_encoded_data.pt"):
    """
    Save the FFT encoded data to a file

    Args:
        train_encoded (dict): Training encoded data
        test_encoded (dict): Testing encoded data
        filename (str): Filename to save to
    """

    data_to_save = {
        'train_encoded': train_encoded,
        'test_encoded': test_encoded,
        'encoding_params': {
            'num_steps': 30,
            'gain': 1,
            'encoding_type': 'rate',
            'data_type': 'fft',
            'frequency_bins': 500,
            'channels': 16
        }
    }

    torch.save(data_to_save, filename)
    print(f"✓ FFT encoded data saved to {filename}")

def load_encoded_fft_data(filename="fft_encoded_data.pt"):
    """
    Load previously encoded FFT data

    Args:
        filename (str): Filename to load from

    Returns:
        tuple: (train_encoded, test_encoded, encoding_params)
    """

    data = torch.load(filename)
    print(f"✓ FFT encoded data loaded from {filename}")
    return data['train_encoded'], data['test_encoded'], data['encoding_params']

# Additional utility functions for FFT-specific analysis

def analyze_frequency_bands(encoded_data, grasp_type, trial_idx=0):
    """
    Analyze spike activity in different frequency bands

    Args:
        encoded_data (dict): Rate-encoded spike data
        grasp_type (str): Grasp type to analyze
        trial_idx (int): Trial index
    """

    trial = encoded_data[grasp_type][trial_idx]  # Shape: (num_steps, 16, 500)

    # Define frequency bands (in terms of bin indices)
    # Assuming 500 bins for 0-1000 Hz: each bin = 2 Hz
    bands = {
        'Low (10-50 Hz)': (5, 25),      # bins 5-25
        'Mid (50-150 Hz)': (25, 75),    # bins 25-75
        'High (150-300 Hz)': (75, 150), # bins 75-150
        'Very High (300-500 Hz)': (150, 250)  # bins 150-250
    }

    print(f"\\n=== FREQUENCY BAND ANALYSIS: {grasp_type} (Trial {trial_idx}) ===")

    for band_name, (start_bin, end_bin) in bands.items():
        band_activity = trial[:, :, start_bin:end_bin]
        avg_activity = band_activity.mean().item()
        total_spikes = band_activity.sum().item()

        print(f"{band_name}:")
        print(f"  - Average spike rate: {avg_activity:.3f}")
        print(f"  - Total spikes: {total_spikes}")
        print(f"  - Relative activity: {avg_activity/trial.mean().item():.2f}x overall average")
        print()
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import LabelEncoder
from collections import Counter
import numpy as np

class FFTSpikeDataset(Dataset):
    def __init__(self, encoded_data):
        """
        Args:
            encoded_data (dict): {'class_name': [Tensor, Tensor, ...], ...}
        """
        self.samples = []
        self.labels = []
        self.classes = list(encoded_data.keys())
        self.encoder = LabelEncoder()
        self.encoder.fit(self.classes)

        for class_name, trials in encoded_data.items():
            label = self.encoder.transform([class_name])[0]
            for trial in trials:
                self.samples.append(trial)
                self.labels.append(label)

        self.samples = torch.stack(self.samples)  # Shape: [N, T, C, F]
        self.labels = torch.tensor(self.labels, dtype=torch.long)

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        return self.samples[idx], self.labels[idx]
```

```python
def create_fft_dataloaders(train_data, test_data, batch_size=32):
    """
    Args:
        train_data (dict): FFT-encoded training data
        test_data (dict): FFT-encoded test data
        batch_size (int): Batch size

    Returns:
        train_loader, test_loader, label_encoder
    """
    train_dataset = FFTSpikeDataset(train_data)
    test_dataset = FFTSpikeDataset(test_data)

    # Ensure label consistency
    test_dataset.encoder = train_dataset.encoder

    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
    test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)

    print(f"Train samples: {len(train_dataset)}")
    print(f"Test samples: {len(test_dataset)}")
    print(f"Classes: {train_dataset.classes}")
    print(f"Train class distribution: {Counter(train_dataset.labels.numpy())}")
    print(f"Test class distribution: {Counter(test_dataset.labels.numpy())}")

    return train_loader, test_loader, train_dataset.encoder
```

```python

class LIFNetwork(nn.Module):
    def __init__(self, input_size, hidden_size, output_size, beta=0.8):
        super(LIFNetwork, self).__init__()

        self.fc1 = nn.Linear(input_size, hidden_size, bias=True)
        self.norm1 = nn.LayerNorm(hidden_size)
        self.lif1 = snn.Leaky(beta=beta, spike_grad=spike_grad)

        self.fc2 = nn.Linear(hidden_size, output_size, bias=True)
        self.norm2 = nn.LayerNorm(output_size)
        self.lif2 = snn.Leaky(beta=beta , spike_grad=spike_grad, output=True)

        # Optional: weight initialization for stability
        nn.init.kaiming_normal_(self.fc1.weight, nonlinearity='linear')
        nn.init.kaiming_normal_(self.fc2.weight, nonlinearity='linear')

    def forward(self, x):
        time_steps, batch_size, _ = x.size()
        mem1 = self.lif1.init_leaky()
        mem2 = self.lif2.init_leaky()

        spk1_rec, spk2_rec = [], []
        mem1_rec, mem2_rec = [], []

        for step in range(time_steps):
            cur1 = self.fc1(x[step])
            cur1 = self.norm1(cur1)               # Added normalization
            spk1, mem1 = self.lif1(cur1, mem1)

            cur2 = self.fc2(spk1)
            cur2 = self.norm2(cur2)               # Added normalization
            spk2, mem2 = self.lif2(cur2, mem2)

            spk1_rec.append(spk1)
            spk2_rec.append(spk2)
            mem1_rec.append(mem1)
            mem2_rec.append(mem2)

        return (torch.stack(spk1_rec), torch.stack(spk2_rec),
                torch.stack(mem1_rec), torch.stack(mem2_rec))


def plot_snn_spikes(spk_in, spk_hidden, spk_out, title="Spiking Neural Network"):
    fig, axes = plt.subplots(3, 1, figsize=(12, 10))

    spk_in_plot = spk_in.squeeze().detach().cpu().numpy()
    if len(spk_in_plot.shape) == 2:
        for i in range(min(spk_in_plot.shape[1], 800)):
            spike_times = np.where(spk_in_plot[:, i] == 1)[0]
            axes[0].scatter(spike_times, [i] * len(spike_times), s=0.5, c='black')
    axes[0].set_title('Input Spikes')
    axes[0].set_ylabel('Input Neuron')
    axes[0].set_xlim(0, spk_in_plot.shape[0])

    spk_hidden_plot = spk_hidden.squeeze().detach().cpu().numpy()
    if len(spk_hidden_plot.shape) == 2:
        for i in range(min(spk_hidden_plot.shape[1], 1000)):
            spike_times = np.where(spk_hidden_plot[:, i] == 1)[0]
            axes[1].scatter(spike_times, [i] * len(spike_times), s=0.5, c='black')
    axes[1].set_title('Hidden Layer')
    axes[1].set_ylabel('Hidden Neuron')
    axes[1].set_xlim(0, spk_hidden_plot.shape[0])

    spk_out_plot = spk_out.squeeze().detach().cpu().numpy()
    if len(spk_out_plot.shape) == 2:
        for i in range(spk_out_plot.shape[1]):
            spike_times = np.where(spk_out_plot[:, i] == 1)[0]
            axes[2].scatter(spike_times, [i] * len(spike_times), s=2, c='black')
    axes[2].set_title('Output Spikes')
    axes[2].set_ylabel('Output Neuron')
    axes[2].set_xlabel('Time Steps')
    axes[2].set_xlim(0, spk_out_plot.shape[0])

    plt.suptitle(title, fontsize=16)
    plt.tight_layout()
    plt.show()

def plot_membrane_traces(mem_rec, spk_rec, title="Membrane Potential Traces"):
    mem_plot = mem_rec.squeeze().detach().cpu().numpy()
    spk_plot = spk_rec.squeeze().detach().cpu().numpy()

    fig, ax = plt.subplots(figsize=(12, 8))
    num_neurons = min(9, mem_plot.shape[1])
    for i in range(num_neurons):
        ax.plot(mem_plot[:, i], label=f'Neuron {i}')
        spike_times = np.where(spk_plot[:, i] == 1)[0]
        spike_values = mem_plot[spike_times, i]
        ax.scatter(spike_times, spike_values, color='red', s=30, zorder=5)

    ax.axhline(y=1.0, color='red', linestyle='--', alpha=0.7, label='Threshold')
    ax.set_xlabel('Time Steps')
    ax.set_ylabel('Membrane Potential')
    ax.set_title(title)
    ax.legend()
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

def run_lif_simulation(train_loader, test_loader=None):
    for batch_data, batch_labels in train_loader:
        batch_size, time_steps, channels, height = batch_data.shape
        break

    input_size = channels * height
    hidden_size = 800
    output_size = 6

    net = LIFNetwork(input_size, hidden_size, output_size, beta=0.7)

    batch_data = batch_data.permute(1, 0, 2, 3)
    batch_data = batch_data.reshape(time_steps, batch_size, -1)

    # Normalize input
    scaler = StandardScaler()
    flat_data = batch_data.reshape(-1, input_size).cpu().numpy()
    scaled_data = scaler.fit_transform(flat_data)
    batch_data = torch.tensor(scaled_data.reshape(time_steps, batch_size, input_size), dtype=torch.float32)

    single_sample = batch_data[:, 0:1, :]

    print(f"Input shape: {single_sample.shape}")
    print(f"Processing through LIF network...")

    with torch.no_grad():
        spk_hidden, spk_out, mem_hidden, mem_out = net(single_sample)

    print(f"Hidden spikes shape: {spk_hidden.shape}")
    print(f"Output spikes shape: {spk_out.shape}")

    plot_snn_spikes(single_sample, spk_hidden, spk_out, "Fully Connected Spiking Neural Network")
    plot_membrane_traces(mem_out, spk_out, "Output Layer Membrane Potential Traces")

    plt.figure(figsize=(12, 6))
    hidden_spike_counts = spk_hidden.squeeze().sum(dim=1).detach().cpu().numpy()
    output_spike_counts = spk_out.squeeze().sum(dim=1).detach().cpu().numpy()
    time_steps_range = range(len(hidden_spike_counts))

    plt.subplot(2, 1, 1)
    plt.plot(time_steps_range, hidden_spike_counts, 'b-', linewidth=2, label='Hidden Layer')
    plt.ylabel('Spike Count')
    plt.title('Spike Activity Over Time')
    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.subplot(2, 1, 2)
    plt.plot(time_steps_range, output_spike_counts, 'r-', linewidth=2, label='Output Layer')
    plt.xlabel('Time Steps')
    plt.ylabel('Spike Count')
    plt.legend()
    plt.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    print(f"\nNetwork Statistics:")
    print(f"Total hidden spikes: {spk_hidden.sum().item()}")
    print(f"Total output spikes: {spk_out.sum().item()}")
    print(f"Hidden spike rate: {spk_hidden.mean().item():.4f}")
    print(f"Output spike rate: {spk_out.mean().item():.4f}")
```

```python
import numpy as np
import os

def apply_fft_to_tensor(tensor, fs=2000, N=2048):
    trials, channels, samples = tensor.shape
    print(f"Input shape: {tensor.shape}")

    fft_output = np.zeros((trials, channels, N//2 + 1))

    for trial_idx in range(trials):
        for channel_idx in range(channels):
            signal = tensor[trial_idx, channel_idx, :N]  # first N samples
            signal -= np.mean(signal)  # remove DC
            fft_result = np.fft.rfft(signal)
            magnitude = np.abs(fft_result)
            fft_output[trial_idx, channel_idx, :] = magnitude

    freqs = np.fft.rfftfreq(N, d=1/fs)
    print(f"FFT output shape: {fft_output.shape}")
    return fft_output, freqs

def process_all_groups_fft(group_folder, output_folder, fs=2000, N=2048):
    group_names = ["G1", "G2", "G3", "G4", "G5", "G6"]
    os.makedirs(output_folder, exist_ok=True)

    freqs_saved = False

    for group in group_names:
        file_path = os.path.join(group_folder, f"{group}.npy")
        if not os.path.exists(file_path):
            print(f"⚠️ Skipping {group}, file not found.")
            continue

        print(f"\n🔁 Processing FFT for: {group}")
        tensor = np.load(file_path)  # shape (trials, 16, 9940)

        fft_tensor, freqs = apply_fft_to_tensor(tensor, fs=fs, N=N)

        # Save FFT tensor
        fft_output_path = os.path.join(output_folder, f"fft_{group}.npy")
        np.save(fft_output_path, fft_tensor)
        print(f"✅ Saved: {fft_output_path}")

# === CONFIG ===
group_folder = "/content/drive/MyDrive/DATASET CSV/processed_groups"
output_folder = "/content/drive/MyDrive/DATASET CSV/fft_outputs"
fs = 2000
N = 1024  # FFT window size

# === EXECUTE ===
process_all_groups_fft(group_folder, output_folder, fs=fs, N=N)
```

```python
import os
import pandas as pd
import numpy as np
from scipy.stats import skew, kurtosis, entropy, mode
from scipy.fft import fft
import csv

# Path setup
parent_folder = '/content/drive/MyDrive/RAW DATA'
output_csv = '/content/drive/MyDrive/DATASET CSV/all_extracted_features.csv'

# Parameters
WAMP_THRESHOLD = 0.01
MYOP_THRESHOLD = 0.05

# Feature extractor per signal (1D numpy array)
def extract_features(signal, sampling_rate=2000):
    signal = np.asarray(signal).flatten()
    N = len(signal)
    diff = np.diff(signal)
    abs_signal = np.abs(signal)
    tke = signal[1:-1]**2 - signal[:-2] * signal[2:]

    # FFT for mean frequency (MF)
    freqs = np.fft.rfftfreq(N, d=1/sampling_rate)
    fft_vals = np.abs(fft(signal))[:len(freqs)]
    mf = np.sum(freqs * fft_vals) / np.sum(fft_vals) if np.sum(fft_vals) != 0 else 0

    # MMAV1 & MMAV2 windowing weights
    n = np.arange(N)
    mmav1_weight = 1 - np.abs((n - N/2)/(N/2))
    mmav2_weight = np.ones(N)
    mmav2_weight[n < 0.25*N] = 0.5
    mmav2_weight[n > 0.75*N] = 0.5

    return {
        # Basic stats
        'mean': np.mean(signal),
        'std_dev': np.std(signal),
        'variance': np.var(signal),
        'max': np.max(signal),
        'min': np.min(signal),
        'median': np.median(signal),
        'range': np.ptp(signal),
        'skewness': skew(signal),
        'kurtosis': kurtosis(signal),

        # Time domain core
        'mav': np.mean(abs_signal),
        'iemg': np.sum(abs_signal),
        'wl': np.sum(np.abs(diff)),
        'zc': np.sum((signal[:-1] * signal[1:] < 0).astype(int)),
        'ssc': np.sum(((np.diff(np.sign(diff))) != 0).astype(int)),
        'wamp': np.sum(np.abs(diff) > WAMP_THRESHOLD),
        'ssi': np.sum(signal**2),
        'dasdv': np.sqrt(np.mean(diff**2)),
        'myop': np.sum(abs_signal > MYOP_THRESHOLD) / N,
        'tke': np.sum(tke),

        # Modified MAVs
        'mmav1': np.sum(abs_signal * mmav1_weight) / N,
        'mmav2': np.sum(abs_signal * mmav2_weight) / N,

        # Frequency domain (mean frequency)
        'mf': mf,

        # Histogram-based
        'hist_mode': mode(signal, keepdims=False).mode,
        'hist_entropy': entropy(np.histogram(signal, bins=32, density=True)[0] + 1e-12)
    }

# Outer function to loop through folders
with open(output_csv, mode='w', newline='') as f:
    writer = None

    for gesture_folder in sorted(os.listdir(parent_folder)):
        gesture_path = os.path.join(parent_folder, gesture_folder)
        if not os.path.isdir(gesture_path): continue

        for filename in os.listdir(gesture_path):
            if filename.endswith('.csv'):
                file_path = os.path.join(gesture_path, filename)
                try:
                    df = pd.read_csv(file_path)
                    all_features = {}
                    rss_sum = 0  # For Root Sum Square (multi-channel)

                    for column in df.columns:
                        if np.issubdtype(df[column].dtype, np.number):
                            signal = df[column].dropna().values
                            feats = extract_features(signal)
                            rss_sum += np.sum(signal**2)
                            for k, v in feats.items():
                                all_features[f"{column}_{k}"] = v

                    all_features['rss'] = np.sqrt(rss_sum)
                    all_features['gesture'] = gesture_folder
                    all_features['file'] = filename

                    if writer is None:
                        writer = csv.DictWriter(f, fieldnames=list(all_features.keys()))
                        writer.writeheader()

                    writer.writerow(all_features)

                except Exception as e:
                    print(f"❌ Error processing {file_path}: {e}")

print("✅ All features extracted and saved to CSV.")
```

```python
import torch
import torch.nn as nn
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from torch.utils.data import TensorDataset, DataLoader
import snntorch as snn
from snntorch import surrogate
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# --------------------- Load and Preprocess Data ---------------------
fft_tensor = np.load("/content/drive/MyDrive/DATASET CSV/fft_outputs/final_fft_tensor_4d.npy")
fft_tensor = torch.tensor(fft_tensor, dtype=torch.float32).reshape(-1, 16, 513)

features_df = pd.read_csv("/content/drive/MyDrive/DATASET CSV/all_extracted_features.csv")
features_df = features_df.drop(columns=["file", "gesture"], errors="ignore")
features = torch.tensor(features_df.values, dtype=torch.float32)

labels = torch.tensor(np.repeat(np.arange(6), 800), dtype=torch.long)

# --------------------- Train-Test Split ---------------------
fft_train, fft_test, feat_train, feat_test, y_train, y_test = train_test_split(
    fft_tensor, features, labels, test_size=0.3, random_state=42, stratify=labels
)

train_dataset = TensorDataset(fft_train, feat_train, y_train)
test_dataset = TensorDataset(fft_test, feat_test, y_test)

train_loader = DataLoader(train_dataset, batch_size=10, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=10, shuffle=False)

# --------------------- Hybrid Model ---------------------
class HybridModel(nn.Module):
    def __init__(self, feature_dim, num_classes):
        super(HybridModel, self).__init__()
        self.conv1 = nn.Conv1d(16, 32, kernel_size=5, stride=1, padding=2)
        self.pool = nn.AvgPool1d(2)
        self.flatten_snn = nn.Flatten()
        self.snn_fc = nn.Linear(32 * (513 // 2), 128)
        self.lif = snn.Leaky(beta=0.95, spike_grad=surrogate.fast_sigmoid())

        self.fc_feat = nn.Sequential(
            nn.Linear(feature_dim, 128),
            nn.ReLU(inplace=False),
        )

        self.fc_combined = nn.Sequential(
            nn.Linear(256, 128),
            nn.ReLU(inplace=False),
            nn.Linear(128, num_classes)
        )

    def forward(self, x_fft, x_feat):
        x_fft = self.conv1(x_fft)
        x_fft = self.pool(x_fft)
        x_fft = x_fft.contiguous().clone()

        x_fft = self.flatten_snn(x_fft)
        x_fft = self.snn_fc(x_fft)
        spk, mem = self.lif(x_fft)

        x_feat = self.fc_feat(x_feat)

        x = torch.cat((mem.detach(), x_feat), dim=1)
        out = self.fc_combined(x)
        return out

# --------------------- Setup ---------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = HybridModel(feature_dim=features.shape[1], num_classes=6).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.CrossEntropyLoss()

# --------------------- Accuracy Function ---------------------
def compute_accuracy(loader):
    model.eval()
    correct = 0
    total = 0
    all_preds = []
    all_labels = []
    with torch.no_grad():
        for fft_batch, feat_batch, label_batch in loader:
            fft_batch = fft_batch.to(device)
            feat_batch = feat_batch.to(device)
            label_batch = label_batch.to(device)
            output = model(fft_batch, feat_batch)
            preds = torch.argmax(output, dim=1)
            correct += (preds == label_batch).sum().item()
            total += label_batch.size(0)
            all_preds.append(preds.cpu())
            all_labels.append(label_batch.cpu())
    return 100.0 * correct / total, torch.cat(all_preds), torch.cat(all_labels)

# --------------------- Training Loop ---------------------
num_epochs = 100
train_accuracies = []
test_accuracies = []

for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0
    for fft_batch, feat_batch, label_batch in train_loader:
        fft_batch = fft_batch.to(device)
        feat_batch = feat_batch.to(device)
        label_batch = label_batch.to(device)

        output = model(fft_batch, feat_batch)
        loss = criterion(output, label_batch)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        running_loss += loss.item()

    train_acc, _, _ = compute_accuracy(train_loader)
    test_acc, _, _ = compute_accuracy(test_loader)

    train_accuracies.append(train_acc)
    test_accuracies.append(test_acc)

    print(f"Epoch {epoch+1}: Loss = {running_loss:.4f}, Train Acc = {train_acc:.2f}%, Test Acc = {test_acc:.2f}%")

# --------------------- Accuracy Plot ---------------------
plt.figure(figsize=(8, 5))
plt.plot(train_accuracies, label="Train Accuracy")
plt.plot(test_accuracies, label="Test Accuracy")
plt.title("Train vs Test Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.legend()
plt.grid(True)
plt.show()

# --------------------- Confusion Matrix ---------------------
_, test_preds, test_targets = compute_accuracy(test_loader)
cm = confusion_matrix(test_targets, test_preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=[f"G{i}" for i in range(6)])
disp.plot(cmap='Blues', xticks_rotation=45)
plt.title("Confusion Matrix (Test Set)")
plt.show()
```

```python
# =================================
# 1. Import Required Libraries
# =================================
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
import matplotlib.pyplot as plt

# =================================
# 2. Load and Prepare Data
# =================================
df = pd.read_csv("/content/drive/MyDrive/DATASET CSV/all_extracted_features.csv")

# Separate features and labels
X = df.drop(columns=['gesture', 'file']).values
y = df['gesture'].values  # Use gesture as target label

# Encode labels
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)
num_classes = len(np.unique(y_encoded))
y_onehot = to_categorical(y_encoded, num_classes)

# =================================
# 3. Normalize and Apply PCA
# =================================
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=10)  # Reduce to 10 principal components
X_pca = pca.fit_transform(X_scaled)
print("Reduced feature shape:", X_pca.shape)

# =================================
# 4. Split Train and Test Data
# =================================
X_train, X_test, y_train, y_test = train_test_split(
    X_pca, y_onehot, test_size=0.2, random_state=42, stratify=y_encoded
)
print("Train shape:", X_train.shape, "Test shape:", X_test.shape)

# =================================
# 5. Build Fully Connected Neural Network
# =================================
feature_dim = X_train.shape[1]  # number of PCA features
model = Sequential([
    Dense(64, activation='relu', input_shape=(feature_dim,)),
    Dropout(0.3),
    Dense(32, activation='relu'),
    Dropout(0.3),
    Dense(num_classes, activation='softmax')
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.summary()

# =================================
# 6. Train the Model
# =================================
history = model.fit(
    X_train, y_train,
    epochs=30,
    batch_size=32,
    validation_split=0.2,
    verbose=1
)

# =================================
# 7. Evaluate on Test Data
# =================================
test_loss, test_accuracy = model.evaluate(X_test, y_test)
print(f"\n✅ Test Accuracy after PCA: {test_accuracy:.4f}")

# =================================
# 8. Plot Training Accuracy
# =================================
plt.figure(figsize=(8, 5))
plt.plot(history.history['accuracy'], label='Train Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title("Training vs Validation Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import torch
import torch.nn as nn
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from torch.utils.data import TensorDataset, DataLoader
import snntorch as snn
from snntorch import surrogate
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# --------------------- Load and Preprocess Data ---------------------
fft_tensor = np.load("/content/drive/MyDrive/DATASET CSV/fft_outputs/final_fft_tensor_4d.npy")
fft_tensor = torch.tensor(fft_tensor, dtype=torch.float32).reshape(-1, 16, 513)

features_df = pd.read_csv("/content/drive/MyDrive/DATASET CSV/pca_reduced_features.csv")
features_df = features_df.drop(columns=["file", "gesture"], errors="ignore")
features = torch.tensor(features_df.values, dtype=torch.float32)

labels = torch.tensor(np.repeat(np.arange(6), 800), dtype=torch.long)

# --------------------- Train-Test Split ---------------------
fft_train, fft_test, feat_train, feat_test, y_train, y_test = train_test_split(
    fft_tensor, features, labels, test_size=0.3, random_state=42, stratify=labels
)

train_dataset = TensorDataset(fft_train, feat_train, y_train)
test_dataset = TensorDataset(fft_test, feat_test, y_test)

train_loader = DataLoader(train_dataset, batch_size=10, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=10, shuffle=False)

# --------------------- Hybrid Model ---------------------
class HybridModel(nn.Module):
    def __init__(self, feature_dim, num_classes):
        super(HybridModel, self).__init__()
        self.conv1 = nn.Conv1d(16, 32, kernel_size=5, stride=1, padding=2)
        self.pool = nn.AvgPool1d(2)
        self.flatten_snn = nn.Flatten()
        self.snn_fc = nn.Linear(32 * (513 // 2), 128)
        self.lif = snn.Leaky(beta=0.95, spike_grad=surrogate.fast_sigmoid())

        self.fc_feat = nn.Sequential(
            nn.Linear(feature_dim, 128),
            nn.ReLU(inplace=False),
        )

        self.fc_combined = nn.Sequential(
            nn.Linear(256, 128),
            nn.ReLU(inplace=False),
            nn.Linear(128, num_classes)
        )

    def forward(self, x_fft, x_feat):
        x_fft = self.conv1(x_fft)
        x_fft = self.pool(x_fft)
        x_fft = x_fft.contiguous().clone()

        x_fft = self.flatten_snn(x_fft)
        x_fft = self.snn_fc(x_fft)
        spk, mem = self.lif(x_fft)

        x_feat = self.fc_feat(x_feat)

        x = torch.cat((mem.detach(), x_feat), dim=1)
        out = self.fc_combined(x)
        return out

# --------------------- Setup ---------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = HybridModel(feature_dim=features.shape[1], num_classes=6).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.CrossEntropyLoss()

# --------------------- Accuracy Function ---------------------
def compute_accuracy(loader):
    model.eval()
    correct = 0
    total = 0
    all_preds = []
    all_labels = []
    with torch.no_grad():
        for fft_batch, feat_batch, label_batch in loader:
            fft_batch = fft_batch.to(device)
            feat_batch = feat_batch.to(device)
            label_batch = label_batch.to(device)
            output = model(fft_batch, feat_batch)
            preds = torch.argmax(output, dim=1)
            correct += (preds == label_batch).sum().item()
            total += label_batch.size(0)
            all_preds.append(preds.cpu())
            all_labels.append(label_batch.cpu())
    return 100.0 * correct / total, torch.cat(all_preds), torch.cat(all_labels)

# --------------------- Training Loop ---------------------
num_epochs = 100
train_accuracies = []
test_accuracies = []

for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0
    for fft_batch, feat_batch, label_batch in train_loader:
        fft_batch = fft_batch.to(device)
        feat_batch = feat_batch.to(device)
        label_batch = label_batch.to(device)

        output = model(fft_batch, feat_batch)
        loss = criterion(output, label_batch)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        running_loss += loss.item()

    train_acc, _, _ = compute_accuracy(train_loader)
    test_acc, _, _ = compute_accuracy(test_loader)

    train_accuracies.append(train_acc)
    test_accuracies.append(test_acc)

    print(f"Epoch {epoch+1}: Loss = {running_loss:.4f}, Train Acc = {train_acc:.2f}%, Test Acc = {test_acc:.2f}%")

# --------------------- Accuracy Plot ---------------------
plt.figure(figsize=(8, 5))
plt.plot(train_accuracies, label="Train Accuracy")
plt.plot(test_accuracies, label="Test Accuracy")
plt.title("Train vs Test Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.legend()
plt.grid(True)
plt.show()

# --------------------- Confusion Matrix ---------------------
_, test_preds, test_targets = compute_accuracy(test_loader)
cm = confusion_matrix(test_targets, test_preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=[f"G{i}" for i in range(6)])
disp.plot(cmap='Blues', xticks_rotation=45)
plt.title("Confusion Matrix (Test Set)")
plt.show()
```
